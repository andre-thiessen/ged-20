## 1 – Introdução
Os dispositivos móveis da série *Moto* geram centenas de métricas operacionais que podem ser exploradas para antecipar problemas de consumo energético. Esta entrega parcial foca em **propor e justificar** uma abordagem de **classificação** capaz de identificar aparelhos com **comportamento energético ineficiente** – isto é, que drenam a bateria de forma anormal mesmo sob um perfil de uso leve ou moderado.  

> **Relevância de negócio**  
> • Reduzir o volume de tickets de suporte relacionados a autonomia de bateria.  
> • Direcionar substituições preventivas de bateria e investigações de firmware.  
> • Fornecer insight para times de engenharia sobre padrões de uso que aceleram o desgaste.  

## 2 – Descrição dos dados
*04/abr/2025* – O arquivo **`Moto_HW_DATA_hashed.csv`** contém os registros anonimizados de telemetria. Cada linha corresponde a **um dispositivo** com valores agregados em nível diário. As colunas de interesse para o nosso projeto (descritas no *Variable_description.xlsx*) são:  

| Variável | Descrição resumida* | Tipo | Observação |
| --- | --- | --- | --- |
| `full_cap_min` | Capacidade mínima medida da bateria (mAh) | Numérica contínua | Valor baixo indica possível desgaste físico. |
| `btlow_total` | Quantidade de alertas de bateria < 15 % | Numérica discreta | Picos sugerem uso em níveis críticos. |
| `dchrg_diff` | Δ % de carga entre ciclos (downloaded charge) | Numérica contínua | > 25 % indica descarga acelerada. |
| `total_chrg_cnt` | Nº total de ciclos de recarga | Numérica discreta | Correlaciona-se com envelhecimento. |
| `son_hrs` | Horas de tela ligada | Numérica contínua | Proxy de intensidade de uso. |
| `msgs_tot` | Nº de mensagens enviadas | Numérica discreta | Tráfego de apps de mensageria. |
| `wifi_hrs` | Horas conectado ao Wi-Fi | Numérica contínua | Alta conectividade ≠ alto consumo. |
| `mobile_rx_MB` | Dados recebidos via rede móvel (MB) | Numérica contínua | Tráfego celular, possível impacto em consumo. |

\*Para a definição completa vide *Variable_description.xlsx*.
